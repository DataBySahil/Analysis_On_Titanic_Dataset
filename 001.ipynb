```python
correlation = df.corr(numeric_only=True)
correlation
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(correlation, annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```

```python
''' 
From the heatmap:

- **Survived and Fare (0.26):** Passengers who paid higher fares had a slightly higher chance of survival.
- **Survived and Pclass (-0.34):** Higher passenger class number (Third Class) is associated with lower survival, meaning first-class passengers had better survival chances.
- **Pclass and Fare (-0.55):** A strong negative correlation indicates that first-class passengers generally paid higher fares.
- **SibSp and Parch (0.41):** Passengers traveling with siblings/spouses were also likely to travel with parents or children.
- **Age and Fare (0.096):** Almost no correlation exists between age and ticket fare.

Overall, the heatmap shows that passenger class and fare had a greater influence on survival than age. 
'''
```
